```python
from pathlib import Path

ROOT = Path(
    "/content/drive/MyDrive/ColabNotebooks/PlayingModels/TaiwanZip"
)

patterns = [
    "*.py",
    "*.ipynb",
    "*.md",
    "*.json",
    "*.yaml",
    "*.yml",
]

files = []

for pattern in patterns:
    files += list(ROOT.rglob(pattern))

for p in sorted(set(files)):
    try:
        size_mb = p.stat().st_size / 1024 / 1024
        print(f"{size_mb:8.2f} MB | {p.relative_to(ROOT)}")
    except:
        pass
```

```python
from pathlib import Path

ROOT = Path(
    "/content/drive/MyDrive/ColabNotebooks/PlayingModels/TaiwanZip"
)

keywords = [
    "spatial_v2",
    "TaiwanZipSpatialVAEV2",
    "regression.npz",
    "latent_channels",
    "semantic_head",
    "depth_head",
    "taiwan_zip_spatial_v2_best",
]

for p in ROOT.rglob("*"):
    if p.suffix.lower() not in [".py", ".ipynb"]:
        continue

    try:
        text = p.read_text(encoding="utf-8", errors="ignore")
    except:
        continue

    hits = [k for k in keywords if k.lower() in text.lower()]

    if hits:
        print("\nFILE:", p.relative_to(ROOT))
        print("HITS:", hits)
```

```python
from pathlib import Path
import re

ROOT = Path(
    "/content/drive/MyDrive/ColabNotebooks/PlayingModels/TaiwanZip"
)

p = ROOT / "taiwan_zip_spatial_vae_v2.py"

text = p.read_text(
    encoding="utf-8",
    errors="ignore"
)

checks = {
    "writes train.npz":
        'train.npz' in text,

    "writes val.npz":
        'val.npz' in text,

    "writes test.npz":
        'test.npz' in text,

    "writes regression.npz":
        'regression.npz' in text,

    "writes dataset_summary.json":
        'dataset_summary.json' in text,

    "creates spatial_v2 bundle":
        'taiwan_zip_spatial_v2_bundle' in text,

    "contains VAE V2":
        'TaiwanZipSpatialVAEV2' in text,

    "writes final checkpoint":
        'taiwan_zip_spatial_v2_best.pt' in text,
}

print("FILE:", p)
print()

for k, v in checks.items():
    print(f"{k:32} : {v}")

print("\n--- FUNCTION DEFINITIONS ---")

for m in re.finditer(r"^def\s+([A-Za-z0-9_]+)\s*\(", text, re.M):
    name = m.group(1)

    if any(word in name.lower() for word in [
        "build",
        "dataset",
        "split",
        "sample",
        "train",
        "save",
        "load",
        "crop",
        "view",
    ]):
        print(name)
```
